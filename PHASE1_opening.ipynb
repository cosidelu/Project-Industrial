```python
from robot_control import RobotController
import Variables as vb
import time

IP_LAB = "192.168.1.3"
controller = RobotController(ip_address=IP_LAB,
                             default_position_j=vb.LOOK_DOWN_POSITION_J_INIZIO)
controller.disconnect()
controller.connect()

SPEED = 400
SPEEDER = 600
SPEEDEST = 800
```

```python
controller.default_positioning()
```

```python
start_time = time.time()
```

## Apertura della visiera

```python
#apertura visiera
controller.move_ptp(vb.VISIERA_0, speed = SPEEDER)
```

```python
controller.move_ptp(vb.VISIERA_1, speed = SPEED)
controller.move_line(vb.VISIERA_2, speed = SPEED)
controller.move_circle(vb.VISIERA_3,vb.VISIERA_4, speed = SPEED)
controller.move_line(vb.VISIERA_5, speed = SPEED)
```

```python
#spostamento per la seconda azione
controller.move_ptp(vb.ALLONTANAMENTO, speed = SPEEDEST)
controller.move_joints(vb.ALLONTANAMENTO_JOINT, speed = SPEEDEST)
```

## Chiusura della visiera oscurata

```python
#chiusura visiera da sole
controller.move_ptp(vb.SOLE_1, speed = SPEEDER)
controller.move_line(vb.SOLE_2, speed = SPEED)
```

```python
controller.move_line(vb.SOLE_3, speed = SPEED)
controller.move_line(vb.SOLE_4, speed = SPEED)
```

## Apertura della visiera oscurata

```python
#pressione pulsante
controller.move_ptp(vb.LUNA_1, speed = SPEED)
controller.move_line(vb.LUNA_2, speed = SPEED)
controller.move_line(vb.LUNA_3, speed = SPEED)
controller.move_line(vb.LUNA_2, speed = SPEED)
```

```python
#più ci allontaniamo e ci prepariamo per la task dopo
controller.move_ptp(vb.LUNA_5, speed = SPEEDER)
controller.move_ptp(vb.LUNA_6, speed = SPEEDER)
```

## Chiusura della visiera

```python
#chiudiamo la visiera
controller.move_joints(vb.VISIERA_10_J, speed = SPEEDER)
controller.move_line(vb.VISIERA_11, speed = SPEED)
controller.move_circle(vb.VISIERA_12,vb.VISIERA_13, speed = SPEED)
controller.move_line(vb.VISIERA_14, speed = SPEED)
```

## Ritorno alla posizione iniziale

Ritorniamo alla posizione iniziale con la stessa configurazione dei joint in modo da garantire ripetibilià ed un passaggio safe alla fase successiva

```python
controller.move_ptp(vb.LOOK_DOWN_POSITION, speed = SPEEDER)
```

```python
controller.default_positioning()
```

```python
end_time = time.time()
elapsed_time = end_time - start_time

print(f"The process took {elapsed_time:.2f} seconds.")
```
